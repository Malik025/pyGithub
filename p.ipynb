## Apply function

##### Define a function that returns
* If marks greater than 80 return A
* If marks greater than 60 return B
* If marks greater than 40 return C
* return fail

```python
def grades(a):
    if a>80:
        return 'A'
    elif a>60:
        return 'B'
    elif a>40:
        return 'C'
    else:
        return 'F'


grades(80)
```

```python
df['Grade'] = df['marks'].apply(grades)
df
```

```python
df['Bonus'] = df['marks'].apply(lambda x:x+5)
df
```

## Join in Python (Merge)

```python
students = pd.DataFrame({'student_id':[1,2,3,6],'name':['Shahzeb','Ammar','Abdullah','Adnan']})
marks = pd.DataFrame({'id':[1,2,3,4],'marks':[100,50,20,50]})

```

```python
d = pd.merge(
    students,
    marks,
    left_on = 'student_id',
    right_on ='id',
    how = 'outer'
)
d[['student_id','marks']]
```
