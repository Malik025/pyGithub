```python
A = np.array([10,20,30,40,50])
print(A>=20)
print(A[A>20])
```

```python
c = np.array([4,9,16,11])
print(np.sqrt(c))
print(np.square(c))

d = np.array([-1,-2])
print(np.abs(d))
```

```python
# Aggregate function:

e = np.array([10,20,30,40])
f = np.array([10,30,20,40,50])

print(np.sum(e))
print(np.max(e))
print(np.min(e))
print(np.mean(e))

print(np.median(e))
print(np.median(f))
```

```python
fg = np.array([10,20,30])

print(np.std(fg)) #Standard deviation
print(np.var(fg)) #Variance
```

### Working with 2d arrays

```python
#axis = 0 -> column
#axis = 1 -> row

arr_2d = np.array([[1,2],[2,3],[4,5]])
print(np.sum(arr_2d,axis=0))
print(np.sum(arr_2d))
print(np.sum(arr_2d,axis=1))
print(np.mean(arr_2d,axis=1))
```
